# CatBoost Regression Pipeline

## Objective

This notebook demonstrates the machine learning workflow used to predict cyclic oxidation mass change using CatBoost regression.

CatBoost was selected because:
- it handles nonlinear relationships effectively
- performs well on tabular scientific datasets
- requires minimal preprocessing
- is robust for small-to-medium sized datasets


```
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

from catboost import CatBoostRegressor
```

```
df = pd.read_csv('../data/Oxi_Cycle.csv')

X = df[['Fe', 'Ni', 'Cr', 'Temperature', 'Time']]
y = df['Mass Change']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```
model = CatBoostRegressor(
    depth=8,
    learning_rate=0.1,
    iterations=1000,
    loss_function='RMSE',
    verbose=False
)

model.fit(X_train, y_train)
```

```
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)

print(f'R²   : {r2:.4f}')
print(f'RMSE : {rmse:.4f}')
print(f'MAE  : {mae:.4f}')
```

## Interpretation

The CatBoost model captures nonlinear oxidation trends reasonably well.

Key observations:
- Ensemble tree-based methods outperform linear models.
- Oxidation behavior depends on coupled effects of composition and temperature.
- Experimental oxidation data contains inherent variability and noise.
