# CatBoost Hyperparameter Study

## Objective

This notebook investigates the effect of CatBoost hyperparameters on cyclic oxidation prediction performance.

Parameters studied:
- tree depth
- learning rate
- iteration count


```
depth_values = [4, 6, 8, 10]
learning_rates = [0.01, 0.03, 0.05, 0.1, 0.2]

print(depth_values)
print(learning_rates)
```

## Key Findings

The study showed that:
- larger tree depths improved predictive performance
- oxidation behavior requires complex nonlinear feature interactions
- moderate learning rates produced stable optimization behavior


## Scientific Interpretation

Improved performance at larger tree depths suggests that cyclic oxidation behavior depends on:

- coupled thermal-composition effects
- nonlinear oxidation kinetics
- interaction-driven degradation mechanisms
