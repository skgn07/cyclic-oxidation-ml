# Cyclic Oxidation Dataset and Materials Background

## Introduction

High-temperature cyclic oxidation is one of the most critical degradation mechanisms affecting structural alloys used in:

- gas turbines
- thermal power plants
- aerospace systems
- heat exchangers
- nuclear applications

Repeated thermal cycling can lead to:
- oxide scale growth
- scale cracking
- spallation
- accelerated material degradation

This notebook introduces the experimental dataset used for machine learning-based cyclic oxidation prediction of Fe–Cr and Fe–Cr–Ni alloy systems.
