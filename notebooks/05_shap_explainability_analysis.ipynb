# SHAP Explainability Analysis

## Objective

Machine learning models should not only predict accurately, but also provide scientific interpretability.

SHAP analysis helps identify:
- dominant features
- feature interactions
- nonlinear effects
- physically meaningful trends


```
import shap
```

## Important Findings

The SHAP analysis indicated that:

- Temperature was the dominant oxidation driver.
- Chromium concentration strongly influenced oxidation resistance.
- Temperature-time interaction effects were important.
- Time alone was less influential than coupled thermal exposure effects.


## Metallurgical Interpretation

These findings are physically meaningful because oxidation kinetics depend strongly on:

- thermal activation
- oxide diffusion
- protective scale formation
- chromium availability
- exposure duration under elevated temperature
