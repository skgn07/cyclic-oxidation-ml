# Alloy-wise Oxidation Analysis

## Objective

This notebook studies oxidation behavior for individual alloy systems.

Analyzing alloy-wise trends is important because:
- different compositions exhibit distinct oxidation kinetics
- binary and ternary systems behave differently
- oxidation resistance depends strongly on chromium content


## Binary Fe–Cr Systems

Binary Fe–Cr alloys primarily demonstrate:
- chromium-controlled oxidation resistance
- oxide scale stabilization
- temperature-dependent degradation


## Ternary Fe–Cr–Ni Systems

Addition of Ni introduces:
- more complex oxidation behavior
- modified diffusion characteristics
- altered oxide scale stability


## Key Observation

The machine learning models were able to reasonably reproduce:
- oxidation growth trends
- mass-change progression
- several spallation-like behaviors
across multiple alloy systems.
