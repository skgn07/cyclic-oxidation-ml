# Model Benchmarking and Comparison

## Objective

This notebook compares multiple machine learning regression models for cyclic oxidation prediction.

Benchmarking is important because:
- different models capture nonlinearities differently
- scientific datasets are often small and noisy
- model selection strongly influences predictive reliability


```
models = [
    'CatBoost Regressor',
    'Extra Trees Regressor',
    'Random Forest Regressor',
    'XGBoost Regressor',
    'Gradient Boosting Regressor',
    'HistGradientBoosting',
    'AdaBoost Regressor',
    'Linear Regression',
    'ElasticNet',
    'KNN Regressor',
    'SVR',
    'MLP Regressor'
]

for model in models:
    print(model)
```

## Key Finding

Tree-based ensemble models significantly outperformed:
- linear regression models
- distance-based models
- shallow learners

This indicates that cyclic oxidation behavior exhibits:
- strong nonlinear dependence
- coupled composition-temperature interactions
- complex oxidation kinetics


## Materials Science Interpretation

Oxidation behavior depends on:
- oxide scale growth
- chromium diffusion
- thermal activation
- oxide spallation
- multivariable interactions

These mechanisms are difficult to capture using simple linear relationships.
